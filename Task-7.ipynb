```python
# Creating a employee class 
class Employee:
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department

#employee id cards      
    def display_id_card(self):
        print("\nEmployee ID Card")
        print(f"Employee Name : {self.name}")
        print(f"Employee ID   : {self.emp_id}")
        print(f"Department    : {self.department}")

#inputs taken by user 
name = input("Employee Name: ")
emp_id = input("Employee ID: ")
department = input("Department: ")

emp = Employee(name, emp_id, department)
emp.display_id_card()
```

```python
# creating class for mobile contacts 
class Contact:

    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

    def display_contact(self):
        print("\nContact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")

#input functions 
name = input("Contact Name: ")
phone = input("Phone Number: ")
contact = Contact(name, phone)
contact.display_contact()
```

```python
#creating class for product 
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display_price_tag(self):
        print("\nProduct:", self.name)
        print(f"Price: {self.price}")

#inputs taking by user's 
name = input("Product Name: ")
price = input("Price: ")
product = Product(name, price)
product.display_price_tag()
```

```python
# creating class movie
class Movie:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating

    def display_movie(self):
        print("\nMovie Details")
        print(f"Movie Name: {self.name}")
        print(f"Rating: {self.rating}")


# input's
name = input("Movie Name: ")
rating = input("Rating: ")

movie = Movie(name, rating)
movie.display_movie()
```

```python
# creating class delivery 
class Delivery:

    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address

    def display_delivery(self):
        print("\nDelivery Details")
        print(f"Customer: {self.customer_name}")
        print(f"Address: {self.address}")

#taking inputs 
customer_name = input("Customer Name: ")
address = input("Address: ")
delivery = Delivery(customer_name, address)
delivery.display_delivery()
```
