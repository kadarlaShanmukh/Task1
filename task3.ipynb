```python
employees = {
"Ravi": 92,
"Anita": 88,
"Kiran": 92,
"Suresh": 85
}
#identifying highest score in give data using max function 
highest_score = max(employees.values())
#creating a list top_performes 
top_performers =[]
#using for loop knowing top performers 
for name,score in employees.items():
# if statement takes, the is equal to highest score 
  if  score == highest_score:
# if the score value is equal to highest value it will append into top_perfomers varaible.
    top_performers.append(name)
#print () will all the required things 
print("top performer eligible for bonus:",",".join(top_performers), f"(score: {highest_score})")
```

```python

s= "Buy mobile phone buy phone online"
# convert string into a lower case 
s= s.lower()
# in string removing a punctuation character 
import string 
for ch in string.punctuation:
    s=s.replace(ch,"")
#it will split the sentence 
words = s.split()
word_count= {}
#using for loop for incrementation of words it existed it will increment automatically 
for word in words:
    if word in word_count:
        word_count[word] +=1
    else:
        word_count[word] = 1
result = {}
for word, count in word_count.items():
    if count>1:
        result[word] = count 
print(result)
```

```python
sensor_readings = [3, 4, 7, 8, 10, 12, 5]
#empty list 
valid_readings = []
# using for loop to calculate every value in the list 
for index, value in enumerate(sensor_readings):
#the value is divide by 2 is equal to zero then append into the valid_readings variable 
    if value % 2 == 0:   # check even number
        valid_readings.append((index, value))
print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)
```

```python
emails = [
"ravi@gmail.com",
"anita@yahoo.com",
"kiran@gmail.com",
"suresh@gmail.com",
"meena@yahoo.com"
]
# taking variables for count 
gmail = 0
yahoo = 0 
for email in emails:
#incrementation of the email  
    if "gmail.com" in email:
        gmail +=1
    elif "yahoo.com" in email:
        yahoo +=1 
total = len(emails)
#percentage formula 
print("gmail.com:", (gmail/total)*100, "%")
print("yahoo.com",(yahoo/total)*100, "%")
```

```python
sales = [1200, 1500, 900, 2200, 1400, 3000]
#Calculate average
average = sum(sales) / len(sales)
print(average)
#30% above average threshold
limit = average * 1.3
for i, value in enumerate(sales, start=1):
    if value > limit:
        print("Day", i, ":", value)
```

```python
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]
count_dict = {}
# Count occurrences
for user in user_ids:
    if user in count_dict:
        count_dict[user] += 1
    else:
        count_dict[user] = 1
# Print duplicates only
for user, count in count_dict.items():
    if count > 1:
        print(user, "→", count, "times")
```
