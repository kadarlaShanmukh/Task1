```python
marks = [45, 78, 90, 33, 60]

#we want to assign variables to for pass and fail students 
PassCount = 0
FailCount = 0
# we will for loop for marks list, to execute no.of iterations & incrementation 
for s in marks:
    if s >=50:
        PassCount +=1
    else: 
        FailCount +=1
# length is know about how many student marks are there 
print(" total length of marks", len(marks))
print(PassCount)
print(FailCount)
       
```

```python
names = [" Alice ", "bob", " CHARLIE "]
#where s is a empty list 
s=[]
for i in names:
#y variable is convert it into a lower case words and remove spaces 
    y=i.strip().lower()
#after convertion we want to add in empty list 
    s.append(y)
# print the s list 
print(s)
```

```python
logs = ["INFO", "ERROR", "WARNING", "ERROR"]
# assigning the varible for error count 
ErrorCount = 0
for i in logs:
    if i == "ERROR":
# +=1 is the incrementation of the couunt ,adding the "Error" count in the variable.
        ErrorCount +=1
print("total error count", ErrorCount)
```
