## An example: visualizing data in the notebook ✨

Below is an example of a code cell. We'll visualize some simple data using two popular packages in Python. We'll use [NumPy](https://numpy.org/) to create some random data, and [Matplotlib](https://matplotlib.org) to visualize it.

Note how the code and the results of running the code are bundled together.

```python
sentence = input("enter the sentence:")
#split the sentence
words = sentence.split()
#creating the empty list 
unique_words = []
for word in words:
    if word not in unique_words:
#using append function to add the word in the list 
        unique_words.append(word)
#Count of unique words 
print("Unique words count:", len(unique_words))
#display the unique words 
print("Unique words:", unique_words)
```

```python
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}
# creating variables 
highest_name = " "
highest_salary = 0
# using for loop for three iterations & and creating variable called name 
for name in employees:
# in name variable if any highest salary person add the data 
    if employees[name] > highest_salary:
        highest_salary = employees[name]
        highest_name = name 
print("highest salary :", highest_name,"-", highest_salary)
```

```python
numbers = [45, 22, 89, 10, 66]

print("List:", numbers)
#using max function and min function from the list 
print("Max:", max(numbers))
print("Min:", min(numbers))
```

```python
prices = [450, 1200, 899, 1500, 300]
# creating empty list 
hike_prize = []
# using for loop if product costs more than 1000 it will insert it into i variable 
for i in prices :
     if i > 1000:
         hike_prize.append(i)
# calculating length of the hike prize 
print("product above 1000:", len(hike_prize))
# we will get output product above 1000
```

```python
attendance = ["P", "P", "A", "P", "P"]
#using the count function to calculate how days present 
s= attendance.count("P")
print( "Count the no.of P:",s)
#total no.of attendance
totalDays = len(attendance)
print("Total no.of days:",totalDays)
#to calculate the atendance percentage 
percentage= s/totalDays*100
print("Attendance Percentage:", percentage)
```

```python
text = "pythonp"
char_count = {}
for s in text:
    if s in char_count:
#incrementation of word if it repeats 
        char_count[s] += 1
    else:
# if it is a new it gives value one 
        char_count[s] = 1
print(char_count)
```
