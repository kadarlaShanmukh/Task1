```python
def validate_recharge(amount):
    valid_plans = [199,299,399,599]
#if plan less than fifty rupess return false 
    if amount < 50:
        print("Recharge amount must be at least 50")
        return False 
# not in plans gives a valid plan structure 
    elif amount not in valid_plans : 
        print (" enter amount not in valid palns: please valid plans like 199, 299, 399, 599")
        return False 
    else:
        return True 
# enter amount in valid plans recharge want to success  
while True:
    try:
        amount = int(input("Enter recharge amount:"))
        if validate_recharge (amount):
            print("Recharge Successfull")
            break
# using except because user gives any input different means it goes to not a valid number like: #50,a50 etc.
    except ValueError:
        print ("please enter a valid number")
```

```python
def check_stock(products):
#using for loop for every item, stock is less than 15 reorder the stock,or stock ok enough 
    for product, stock in products.items():
        if stock < 15:
            print(product, "- Reorder Alert")
        else:
            print(product, "- Stock OK")
products = {
    "Rice": 20,
    "Oil": 10,
    "Sugar": 25,
    "Salt": 12
}
check_stock(products)
```

```python
# Function to calculate result
def check_result(marks):
    total = 0
# Loop to calculate total
    for mark in marks:
        total += mark
    average = total / len(marks)
    if average >= 50:
        result = "Pass"
    else:
        result = "Fail"
    
    return average, result
# Example list of marks
student_marks = [60, 70, 50, 40]
avg, res = check_result(student_marks)
print("Average Marks:", avg)
print("Result:", res)
```

```python
# Function to calculate fare
def calculate_fare(distance, peak):
    base_fare = 50
    per_km = 12
# calculation for without peak hours 
    total = base_fare + (distance * per_km)
#when was peak hours add 25% extra to the per kilometer charges 
    if peak.lower() == "yes":
        total = total + (0.25 * total)   # 25% extra
    return total

while True:
    try:
        distance = float(input("Enter distance in km: "))
        if distance <= 0:
            print("Distance must be greater than 0!")
            continue
        peak = input("Is it peak hour? (yes/no): ")
        fare = calculate_fare(distance, peak)
        print("Total Fare: ₹", fare)
        break   
# stop after successful calculation
    except ValueError:
        print("Please enter valid number for distance!")
```

```python
def check_attendance(attendance_list):
    present_days = 0

# using incrementation for using counting dates 
    for day in attendance_list:
        if day == "P":
            present_days += 1
    total_days = len(attendance_list)
# Avoid division by zero
    if total_days == 0:
        return "No Data"
    percentage = (present_days / total_days) * 100
    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"
# Example usage
attendance = ["P", "A", "P", "P", "A", "P", "P", "P"]
result = check_attendance(attendance)
print(result)
```

```python
def check_password(password):
#need password atleast 8 characters 
    if len(password) < 8:
        return "Password is Weak (Minimum 8 characters required)"

    has_digit = False
    has_special = False
# Loop through each character
    for ch in password:
        if ch.isdigit():
            has_digit = True
        if ch in "@#$":
            has_special = True
# Final check
    if has_digit and has_special:
        return "Password is Strong ✅"
    else:
        return "Password is Weak (Must contain digit & special character)"
# Taking input
pwd = input("Enter your password: ")
result = check_password(pwd)
print(result)
```
