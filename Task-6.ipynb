```python
def smart_parking(capacity, vehicle_logs):
    parked = 0
    peak_usage = 0

    for log in vehicle_logs:
        if log == "IN":
            parked += 1
        elif log == "OUT":
            parked -= 1
# Track peak usage
        if parked > peak_usage:
            peak_usage = parked
# Alert if capacity exceeded
        if parked > capacity:
            print(" Alert: Parking capacity exceeded!")
# Parking status
    if parked < capacity:
        status = "Available"
    else:
        status = "Full"

    print("Currently Parked Vehicles:", parked)
    print("Peak Parking Usage:", peak_usage)
    print("Parking Status:", status)
# Sample Input
capacity = 50
vehicle_logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]
# Function Call
smart_parking(capacity, vehicle_logs)
```

```python
def estimate_delivery_time(distance, traffic, weather):
# Base time 5 minutes per km
    delivery_time = distance * 5
# Traffic delay prizes 
    if traffic.lower() == "high":
        delivery_time += 10
    elif traffic.lower() == "medium":
        delivery_time += 5
    elif traffic.lower() == "low":
        delivery_time += 0
# Weather delay extra priz
    if weather.lower() == "rainy":
        delivery_time += 5
    elif weather.lower() == "stormy":
        delivery_time += 15
    elif weather.lower() == "clear":
        delivery_time += 0
    return delivery_time
    
distance = int(input("Distance (km): "))
traffic = input("Traffic Level: ")
weather = input("Weather: ")

s = estimate_delivery_time(distance, traffic, weather)
print(f"Estimated Delivery Time: {s} minutes")
```

```python
def seat_occupancy_analyzer(total_seats, booked_seats):
# Count booked seats
    booked_count = sum(booked_seats)
# Calculate occupancy %
    occupancy = (booked_count / total_seats) * 100
# Decide show status
    if occupancy >= 90:
        status = "Houseful"
    elif occupancy >= 70:
        status = "Almost Full"
    elif occupancy >= 40:
        status = "Moderate"
    else:
        status = "Available"
    return occupancy, status
    
total_seats = 200
# 150 seats booked
booked_seats = [1] * 150 
occupancy, status = seat_occupancy_analyzer(total_seats, booked_seats)
print(f"Occupancy: {occupancy:}%")
print(f"Show Status: {status}")
```

```python
def classify_svr_load(cpu_readings):
# Calculate average CPU load
    avg_load = sum(cpu_readings) / len(cpu_readings)
# Classification
    if avg_load < 50:
        status = "Normal"
    elif avg_load <= 80:
        status = "Warning"
    else:
        status = "Critical"
    return avg_load, status
#call to action 
cpu_readings = [45, 60, 70, 85, 90]
average, server_status = classify_svr_load(cpu_readings)
print(f"Average CPU Load: {average:}")
print(f"Server Status: {server_status}")
```

```python
def resource_usage_monitor(resources):
    overused = []
# Check each resource usage
    for resource, hours in resources.items():
        if hours > 8:
            overused.append(resource)
# Energy alert condition
    if overused:
        alert = "Yes"
    else:
        alert = "No"
    return overused, alert

#resource 
resource_usage = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

#CallToAction
overused_resources, energy_alert = resource_usage_monitor(resource_usage)
print("Overused Resources:", ", ".join(overused_resources))
print("Energy Alert:", energy_alert)
```

```python
def event_registration_controller(capacity, registrations):
# Check capacity
    if registrations <= capacity:
        confirmed = registrations
        waitlisted = 0
        status = "Open"
    else:
        confirmed = capacity
        waitlisted = registrations - capacity
        status = "Closed"

    return confirmed, waitlisted, status

# inputs 
event_capacity = 100
registrations = 105

#calltoaction
confirmed, waitlisted, reg_status = event_registration_controller(event_capacity, registrations)

print(f"Confirmed Registrations: {confirmed}")
print(f"Waitlisted Users: {waitlisted}")
print(f"Registration Status: {reg_status}")
```
