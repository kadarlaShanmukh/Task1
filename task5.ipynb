```python
# Function to analyze engagement
def analyze_engagement(likes_list):
        
    total_likes = 0
# Using loop for calculations
    for like in likes_list:
        total_likes += like
# Check post status
    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"
    return total_likes, status
# Example likes data
likes = [200, 150, 300, 250, 150]
# Function call
total, result = analyze_engagement(likes)
# Output
print("Total Likes:", total)
print("Post Status:", result)
```

```python
# Function to check medicine stock
def check_stock(stock):
# Conditional logic
    if stock < 10:
        status = "Low Stock Alert"
    else:
        status = "Stock Sufficient"
    return status
# Example stock value
medicine_stock = 6
# Function call
result = check_stock(medicine_stock)
# Output
print("Medicine Stock:", medicine_stock)
print("Status:", result)
```

```python
# Function to analyze rainfall
def analyze_rainfall(rainfall_data, required_level):

    total = 0
# Using loop to calculate total rainfall
    for rain in rainfall_data:
        total += rain
# Calculate average rainfall
    average = total / len(rainfall_data)
# Check rainfall status
    if average >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"

    return average, status
# Example rainfall data
rainfall = [70, 75, 68, 80, 67]
required = 70
# Function call
avg, result = analyze_rainfall(rainfall, required)
# Output
print("Average Rainfall:", int(avg))
print("Rainfall Status:", result)
```

```python
# Function to detect duplicate usernames
def check_duplicates(usernames):
# Convert list into set
    unique_users = set(usernames)
# Check duplicates
    if len(unique_users) < len(usernames):
        result = "Yes"
    else:
        result = "No"

    return result
# Example usernames
users = ["john", "alice", "mark", "john", "david"]
# Function call
output = check_duplicates(users)
# Output
print("Duplicate Accounts Found:", output)
```

```python
# Function to check eligibility
def check_eligibility(age):
# Condition check
    if age >= 18:
        status = "Eligible"
    else:
        status = "Not Eligible"

    return status
# Example patient age
patient_age = 21
# Function call
result = check_eligibility(patient_age)
# Output
print("Patient Age:", patient_age)
print("Eligibility Status:", result)
```

```python
# Function to filter premium crops
def filter_premium_crops(prices):

    premium_crops = []
# List filtering using loop
    for price in prices:
        if price > 2000:
            premium_crops.append(price)

    return premium_crops
# Example crop prices
crop_prices = [1500, 2500, 1800, 3200, 900]
# Function call
result = filter_premium_crops(crop_prices)
# Output
print("Premium Crops:", result)
```

```python
# Function to evaluate application health
def check_system_health(errors):

    # Conditional statements
    if errors == 0:
        status = "Healthy"
    elif errors <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"

    return status
# Example error count
error_count = 7
# Function call
result = check_system_health(error_count)
# Output
print("Error Count:", error_count)
print("System Status:", result)
```

```python
# Function to validate transaction amount
def validate_transaction(amount):

    daily_limit = 50000
# Conditional logic
    if amount <= daily_limit:
        status = "Approved"
    else:
        status = "Rejected"
    return status
# Example transaction amount
transaction_amount = 60000
# Function call
result = validate_transaction(transaction_amount)
# Output
print("Transaction Amount:", transaction_amount)
print("Transaction Status:", result)
```

```python
# Function to calculate attendance and eligibility
def check_attendance(attendance_list):

    total_classes = len(attendance_list)
    attended_classes = 0
# Using loop to count attended days
    for day in attendance_list:
        if day == 1:
            attended_classes += 1
# Calculate attendance percentage
    percentage = (attended_classes / total_classes) * 100
# Check eligibility
    if percentage >= 75:
        status = "Eligible"
    else:
        status = "Not Eligible"

    return percentage, status
# Example attendance (1 = Present, 0 = Absent)
attendance = [1, 1, 1, 0, 1]
# Function call
percent, result = check_attendance(attendance)
# Output
print("Attendance Percentage:", percent)
print("Exam Eligibility:", result)
```

```python
# Function to calculate electricity bill
def calculate_bill(units):
# Billing calculation using slabs
    if units <= 100:
        bill = units * 3

    elif units <= 200:
        bill = (100*3)+(units-100)*5

    else:
        bill = (100 * 3)+(100 * 5)+(units - 200)*7
# Usage classification
    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"

    return bill, usage
# Example units consumed
units_used = 250
# Function call
total_bill, usage_status = calculate_bill(units_used)
# Output
print("Units Consumed:", units_used)
print("Total Bill: ₹", total_bill)
print("Usage Category:", usage_status)
```
